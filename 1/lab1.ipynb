```python
#imports
import matplotlib.pyplot as plt
import numpy as np
from statistics import mean, stdev
```

```python
#constants
alpha = 190
lamb = 1/130
a = 80

size = [100, 1000, 10000]
uni = []
exp = []
treug = []
```

1. Равномерное

```python
for i in size:
  #dataset
  uni_data = np.random.uniform(0, alpha, i)
  uni.append(uni_data)

  #plot
  plot = plt.hist(uni_data, bins=50, density=True)
  plt.title(f"Равномерное распределение ({i} элементов)")
  plt.xlabel("Значение X")
  plt.ylabel("Плотность")
  plt.grid()
  plt.show()
```

```python
#parameters
mat = alpha/2
sko = (alpha**2/12)**(1/2)
print(f"Мат. ожидание (теор.): {mat}, СКО (теор.): {sko}\n")

for i in range(3):
  mat_prac = mean(uni[i])
  sko_prac = stdev(uni[i])
  print(f"Для выборки из {size[i]} элементов")
  print(f"Мат. ожидание (практ.): {mat_prac}, СКО (практ.): {sko_prac}\n")
```

2. Экспоненциальное

```python
for i in size:
  #dataset
  u = np.random.random(i)
  exp_data = (-1 / lamb) * np.log(1 - u)
  exp.append(exp_data)

  #plot
  plt.hist(exp_data, bins=50, density=True)
  plt.title(f"Экспоненциальное распределение ({i} элементов)")
  plt.xlabel("Значение X")
  plt.ylabel("Плотность")
  plt.grid()
  plt.show()
```

```python
#parameters
mat = 1/lamb
sko = 1/lamb
print(f"Мат. ожидание (теор.): {mat}, СКО (теор.): {sko}\n")

for i in range(3):
  mat_prac = mean(exp[i])
  sko_prac = stdev(exp[i])
  print(f"Для выборки из {size[i]} элементов")
  print(f"Мат. ожидание (практ.): {mat_prac}, СКО (практ.): {sko_prac}\n")
```

3. Треугольное

```python
for i in size:
  #dataset
  u = np.random.random(10000)
  treug_data = a * (1 - np.sqrt(u))
  treug.append(treug_data)

  #plot
  plt.hist(treug_data, bins=50, density=True)
  plt.title(f"Треугольное распределение ({i} элементов)")
  plt.xlabel("Значение X")
  plt.ylabel("Плотность")
  plt.grid()
  plt.show()
```

```python
#parameters
mat = a/3
sko = (a**2/18)**(1/2)
print(f"Мат. ожидание (теор.): {mat}, СКО (теор.): {sko}\n")

for i in range(3):
  mat_prac = mean(treug[i])
  sko_prac = stdev(treug[i])
  print(f"Для выборки из {size[i]} элементов")
  print(f"Мат. ожидание (практ.): {mat_prac}, СКО (практ.): {sko_prac}\n")
```
